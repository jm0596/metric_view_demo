```python
%sql

CREATE OR REPLACE VIEW metric_view_demo.my_schema.order_metrics_sql
WITH METRICS
LANGUAGE YAML
AS $$
  version: 1.1
  source: samples.tpch.orders
  comment: Order metrics with revenue and order counts, slicable by date and year
  dimensions:
    - name: order_date
      expr: o_orderdate
      synonyms:
        - purchase date
        - transaction date
    - name: order_year
      expr: YEAR(o_orderdate)
  measures:
    - name: total_revenue
      display_name: Total Revenue (USD)
      expr: SUM(o_totalprice)
      format:
        type: currency
        currency_code: USD
        decimal_places:
          type: exact
          places: 2
        abbreviation: compact
        hide_group_separator: false

      synonyms:
        - revenue
        - sales
        - gross revenue
    - name: order_count
      display_name: Order Count
      expr: COUNT(*)
$$
```
